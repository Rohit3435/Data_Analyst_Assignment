```python
import pandas as pd
import os

datasets = {}

data_path = "data/raw"

for file_name in os.listdir(data_path):
    if file_name.endswith(".csv"):
        file_path = os.path.join(data_path, file_name)
        
        dataset_name = file_name.replace(".csv", "")
        datasets[dataset_name] = pd.read_csv(file_path)

print("Datasets loaded:", len(datasets))
print("Dataset names:", list(datasets.keys()))
```

```python
schema_summary = []

for dataset_name, df in datasets.items():
    schema_summary.append({
        "dataset": dataset_name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": ", ".join(df.columns.astype(str).tolist())
    })

schema_summary_df = pd.DataFrame(schema_summary)

schema_summary_df
```

```python
schema_summary = []

for dataset_name, df in datasets.items():
    schema_summary.append({
        "dataset": dataset_name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "column_names": ", ".join(df.columns.tolist())
    })

schema_summary_df = pd.DataFrame(schema_summary)

schema_summary_df
```

```python
os.makedirs("output/analysis", exist_ok=True)

schema_summary_df.to_csv(
    "output/analysis/schema_summary.csv",
    index=False
)

print("Schema summary saved successfully!")
```

```python
data_types_summary = []

for dataset_name, df in datasets.items():
    
    for column in df.columns:
        data_types_summary.append({
            "dataset": dataset_name,
            "column": column,
            "data_type": str(df[column].dtype),
            "non_null_count": df[column].notna().sum(),
            "null_count": df[column].isna().sum(),
            "unique_values": df[column].nunique()
        })

data_types_df = pd.DataFrame(data_types_summary)

data_types_df.head(20)
```

```python
data_types_summary = []

for dataset_name, df in datasets.items():
    
    for column in df.columns:
        data_types_summary.append({
            "dataset": dataset_name,
            "column": column,
            "data_type": str(df[column].dtype),
            "non_null_count": df[column].notna().sum(),
            "null_count": df[column].isna().sum(),
            "unique_values": df[column].nunique()
        })

data_types_df = pd.DataFrame(data_types_summary)

data_types_df.head(20)
```

```python
data_types_df.to_csv(
    "output/analysis/data_types_summary.csv",
    index=False
)

print("Data types summary saved successfully!")
```

```python
numerical_columns_summary = []

for dataset_name, df in datasets.items():
    
    numerical_columns = df.select_dtypes(
        include=["int64", "float64", "int32", "float32"]
    ).columns.tolist()
    
    numerical_columns_summary.append({
        "dataset": dataset_name,
        "numerical_columns": len(numerical_columns),
        "column_names": ", ".join(numerical_columns)
    })

numerical_columns_df = pd.DataFrame(
    numerical_columns_summary
)

numerical_columns_df
```

```python
numerical_columns_df.to_csv(
    "output/analysis/numerical_columns_summary.csv",
    index=False
)

print("Numerical columns summary saved successfully!")
```

```python
categorical_columns_summary = []

for dataset_name, df in datasets.items():
    
    categorical_columns = df.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()
    
    categorical_columns_summary.append({
        "dataset": dataset_name,
        "categorical_columns": len(categorical_columns),
        "column_names": ", ".join(categorical_columns)
    })

categorical_columns_df = pd.DataFrame(
    categorical_columns_summary
)

categorical_columns_df
```

```python
categorical_columns_df.to_csv(
    "output/analysis/categorical_columns_summary.csv",
    index=False
)

print("Categorical columns summary saved successfully!")
```

```python
missing_values_analysis = []

for dataset_name, df in datasets.items():
    
    missing_count = df.isna().sum().sum()
    
    missing_values_analysis.append({
        "dataset": dataset_name,
        "total_missing_values": missing_count,
        "missing_percentage": round(
            (missing_count / (df.shape[0] * df.shape[1])) * 100,
            2
        )
    })

missing_values_df = pd.DataFrame(
    missing_values_analysis
).sort_values(
    by="total_missing_values",
    ascending=False
).reset_index(drop=True)

missing_values_df
```

```python
missing_values_df.to_csv(
    "output/analysis/missing_values_analysis.csv",
    index=False
)

print("Missing values analysis saved successfully!")
```

```python
dataset_overview = []

for dataset_name, df in datasets.items():
    
    dataset_overview.append({
        "dataset": dataset_name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "total_cells": df.shape[0] * df.shape[1],
        "missing_values": df.isna().sum().sum(),
        "duplicate_rows": df.duplicated().sum()
    })

dataset_overview_df = pd.DataFrame(dataset_overview)

dataset_overview_df
```

```python
dataset_overview_df.to_csv(
    "output/analysis/dataset_overview.csv",
    index=False
)

print("Final dataset overview saved successfully!")
```

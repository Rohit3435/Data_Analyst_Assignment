```python
import pandas as pd
import os
from pathlib import Path

# ==============================
# FINAL INSIGHTS & RECOMMENDATIONS
# ==============================

golden_path = "output/golden_dataset"

print("=" * 70)
print("FINAL BUSINESS INSIGHTS")
print("=" * 70)

# Load all golden datasets
datasets = {}

for file_name in os.listdir(golden_path):
    if file_name.endswith(".csv"):
        file_path = os.path.join(golden_path, file_name)
        dataset_name = file_name.replace("_golden.csv", "")
        datasets[dataset_name] = pd.read_csv(file_path)

print(f"\nGolden datasets loaded: {len(datasets)}")

# ------------------------------
# DATASET OVERVIEW
# ------------------------------

summary = []

for name, df in datasets.items():
    summary.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_values": int(df.isnull().sum().sum()),
        "duplicates": int(df.duplicated().sum())
    })

summary_df = pd.DataFrame(summary)

print("\nDATASET QUALITY SUMMARY")
print(summary_df.sort_values("rows", ascending=False))

# ------------------------------
# KEY BUSINESS INSIGHTS
# ------------------------------

print("\n" + "=" * 70)
print("KEY BUSINESS INSIGHTS")
print("=" * 70)

print(f"""
1. A total of {len(datasets)} datasets were successfully processed into the Golden Dataset layer.

2. The largest datasets contain the highest volume of operational activity and should receive priority monitoring.

3. Data quality issues such as duplicate records were handled during Golden Dataset creation.

4. Missing-value patterns and duplicate records should continue to be monitored through automated validation.

5. The Golden Dataset provides a clean and standardized foundation for reporting, KPI tracking, and business decision-making.
""")

# ------------------------------
# FINAL RECOMMENDATIONS
# ------------------------------

print("\n" + "=" * 70)
print("FINAL RECOMMENDATIONS")
print("=" * 70)

recommendations = [
    "Implement automated data-quality checks before new data enters the reporting pipeline.",
    "Monitor duplicate records regularly to prevent inaccurate business reporting.",
    "Track missing values for critical business fields and investigate recurring issues.",
    "Use the Golden Dataset as the single trusted source for analytics and dashboards.",
    "Build KPI dashboards on top of the cleaned datasets for continuous business monitoring."
]

for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")

# ------------------------------
# SAVE FINAL SUMMARY
# ------------------------------

os.makedirs("output/final_analysis", exist_ok=True)

summary_df.to_csv(
    "output/final_analysis/final_dataset_summary.csv",
    index=False
)

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)
print("\nFinal summary saved to:")
print("output/final_analysis/final_dataset_summary.csv")
```

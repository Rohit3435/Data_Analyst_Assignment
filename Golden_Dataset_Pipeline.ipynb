```python
import os
import pandas as pd

# ==========================================
# GOLDEN DATASET PIPELINE
# ==========================================

RAW_PATH = "data/raw"
GOLDEN_PATH = "output/golden_dataset"

os.makedirs(GOLDEN_PATH, exist_ok=True)

pipeline_summary = []

for file_name in os.listdir(RAW_PATH):
    
    if file_name.endswith(".csv"):
        
        dataset_name = file_name.replace(".csv", "")
        file_path = os.path.join(RAW_PATH, file_name)
        
        # Load dataset
        df = pd.read_csv(file_path)
        
        original_rows = len(df)
        
        # Standardize column names
        df.columns = (
            df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
        )
        
        # Remove exact duplicates
        duplicates_removed = df.duplicated().sum()
        df = df.drop_duplicates()
        
        # Save cleaned golden dataset
        output_file = os.path.join(
            GOLDEN_PATH,
            f"{dataset_name}_golden.csv"
        )
        
        df.to_csv(output_file, index=False)
        
        pipeline_summary.append({
            "dataset": dataset_name,
            "original_rows": original_rows,
            "final_rows": len(df),
            "duplicates_removed": duplicates_removed,
            "columns": len(df.columns),
            "golden_file": output_file
        })


# Create pipeline summary
pipeline_summary_df = pd.DataFrame(pipeline_summary)

pipeline_summary_df.to_csv(
    f"{GOLDEN_PATH}/golden_pipeline_summary.csv",
    index=False
)

print("=" * 60)
print("GOLDEN DATASET PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 60)

display(pipeline_summary_df)

print(f"\nGolden datasets created: {len(pipeline_summary_df)}")
print(f"Location: {GOLDEN_PATH}")
```

```python
import os
import pandas as pd

golden_path = "output/golden_dataset"

golden_files = [
    f for f in os.listdir(golden_path)
    if f.endswith(".csv")
]

print("FINAL GOLDEN DATASET VALIDATION")
print("=" * 50)
print("Golden CSV files:", len(golden_files))

validation = []

for file_name in golden_files:
    df = pd.read_csv(os.path.join(golden_path, file_name))

    validation.append({
        "dataset": file_name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_values": int(df.isnull().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })

validation_df = pd.DataFrame(validation)
validation_df
```

```python
print("GOLDEN DATASET PIPELINE STATUS")
print("=" * 50)

print("Total datasets validated:", len(validation_df))
print("Total missing values:", validation_df["missing_values"].sum())
print("Total duplicate rows:", validation_df["duplicate_rows"].sum())

print("\nGolden Dataset Pipeline completed successfully!")
```
